```python
#a) Piste liikkuu yksikköympyrällä kulmanopeudella ω=2 rad/s. 
# Piirrä Pythonilla sen koordinaatit x(t)=cos(2t) ja y(t)=sin(2t) ajan funktiona.

#Määritellään haluttu aikaväli t:
t = np.arange(0,10,0.01)
#Lasketaan x ja y
x = np.cos(2*t)
y = np.sin(2*t)

#Piirretään kuva
plt.figure(figsize=(10,3))
plt.plot(t,x)
plt.plot(t,y)
plt.legend(['x','y'])
plt.grid()
```

```python
#7 b) Piirrä f(t)=3sin(2t+1). Muodosta kosinifunktio, jonka kuvaaja on sama.
#Aika
t = np.arange(0,10,0.01)

#Lasketaan f
f = 3*np.sin(2*t + 1)

#Piirretään kuva
plt.figure(figsize=(10,3))
plt.plot(t,f)
plt.grid()

#Muodosta kosinifunktio, jonka kuvaaja on sama.
#Yleisesti voimme todeta, että sinin ja kosinen välinen vaihe-ero on pi/2
g = 3*np.cos(2*t + 1 - np.pi/2)
plt.plot(t,g)
```

```python
#c) Määritä funktion  2/3*sin(7t−1) amplitudi, kulmataajuus, vaihe ja jaksonaika.
#Aika
t = np.arange(0,5,0.01)

#Lasketaan f
f = 2/3*np.sin(7*t - 1)

#Piirretään kuva
plt.figure(figsize=(10,3))
plt.plot(t,f)
plt.grid()
```

```python
#d) Muodosta sinifunktio, jonka amplitudi on 3, jaksonaika 4 s ja vaihe π.
#Aika
t = np.arange(0,10,0.01)

#Lasketaan f
A = 3
omega = np.pi/2
phi = np.pi
f = A*np.sin(omega*t + phi)

#Piirretään kuva
plt.figure(figsize=(10,3))
plt.plot(t,f)
plt.grid()
```

```python
#8 a) Määritä kuvaajasta amplitudi, jaksonaika ja vaihe. Muodosta kuvan perusteella funktio ja vertaa sen Python-kuvaajaa alkuperäiseen.
#Aika
t = np.arange(0,10,0.01)

#Lasketaan f
A = 2.5
omega = 1.01*np.pi
phi = 0
f = A*np.sin(omega*t + phi)

#Piirretään kuva
plt.figure(figsize=(10,3))
plt.plot(t,f)
plt.grid()
```
